```python
# Number of users
NUM_USERS = 100000

# For reproducibility
np.random.seed(42)

# Create User IDs
users = pd.DataFrame({
    "user_id": [
        f"U{str(i).zfill(6)}"
        for i in range(1, NUM_USERS + 1)
    ]
})

# Assign device
users["device"] = np.random.choice(
    ["Mobile", "Desktop", "Tablet"],
    size=NUM_USERS,
    p=[0.60, 0.35, 0.05]
)

# Assign location
users["location"] = np.random.choice(
    [
        "Mumbai",
        "Pune",
        "Delhi",
        "Bangalore",
        "Hyderabad",
        "Chennai",
        "Kolkata",
        "Ahmedabad"
    ],
    size=NUM_USERS
)

# Assign traffic source
users["traffic_source"] = np.random.choice(
    [
        "Google",
        "Instagram",
        "Facebook",
        "YouTube",
        "Email",
        "Direct",
        "Referral",
        "Paid Ads"
    ],
    size=NUM_USERS,
    p=[0.20, 0.18, 0.12, 0.08, 0.10, 0.12, 0.08, 0.12]
)

# Assign customer type
users["customer_type"] = np.random.choice(
    ["New", "Returning"],
    size=NUM_USERS,
    p=[0.75, 0.25]
)

print("Users generated:", len(users))
```

```python
# Generate number of sessions for each user
session_counts = np.random.choice(
    [1, 2, 3, 4],
    size=NUM_USERS,
    p=[0.55, 0.25, 0.15, 0.05]
)

print("Total users:", NUM_USERS)
print("Average sessions per user:", session_counts.mean())
print("Maximum sessions for one user:", session_counts.max())
```

```python
sessions = []

session_id = 1

START_DATE = "2025-01-01"
END_DATE = "2025-06-30"

start_date = pd.Timestamp(START_DATE)
end_date = pd.Timestamp(END_DATE)

for user, count in zip(users.itertuples(index=False), session_counts):

    for _ in range(count):

        # Random date between start and end date
        session_date = start_date + pd.Timedelta(
            days=np.random.randint(
                0,
                (end_date - start_date).days + 1
            )
        )

        sessions.append({
            "session_id": f"S{session_id:07d}",
            "user_id": user.user_id,
            "session_date": session_date,
            "device": user.device,
            "location": user.location,
            "traffic_source": user.traffic_source,
            "customer_type": user.customer_type
        })

        session_id += 1

# Convert list to DataFrame
sessions = pd.DataFrame(sessions)

print("Total sessions:", len(sessions))
```

```python
# Product categories and products

products = {
    "Electronics": [
        "Laptop",
        "Smartphone",
        "Headphones",
        "Tablet"
    ],

    "Fashion": [
        "Shoes",
        "T-Shirt",
        "Jeans",
        "Jacket"
    ],

    "Home": [
        "Chair",
        "Table",
        "Lamp",
        "Bed"
    ],

    "Beauty": [
        "Perfume",
        "Skincare",
        "Makeup",
        "Haircare"
    ],

    "Sports": [
        "Running Shoes",
        "Cricket Bat",
        "Football",
        "Gym Equipment"
    ]
}

categories = list(products.keys())

print("Categories:", categories)
```

```python
price_ranges = {

    "Electronics": (5000, 100000),

    "Fashion": (500, 10000),

    "Home": (1000, 30000),

    "Beauty": (300, 8000),

    "Sports": (500, 20000)
}

price_ranges
```

```python
# Device conversion multiplier

device_multiplier = {
    "Mobile": 0.82,
    "Desktop": 1.00,
    "Tablet": 0.75
}


# Traffic source conversion multiplier

source_multiplier = {
    "Google": 1.05,
    "Instagram": 0.75,
    "Facebook": 0.70,
    "YouTube": 0.80,
    "Email": 1.20,
    "Direct": 1.05,
    "Referral": 0.95,
    "Paid Ads": 0.85
}


# Customer type multiplier

customer_multiplier = {
    "New": 0.85,
    "Returning": 1.20
}
```

```python
events = []

event_id = 1

for row in sessions.itertuples(index=False):

    # Random time during the session day
    base_time = row.session_date + pd.Timedelta(
        minutes=np.random.randint(0, 1440)
    )

    # --------------------------------
    # STAGE 1: VISIT
    # --------------------------------

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": base_time,
        "event_type": "visit",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": None,
        "product": None,
        "amount": 0
    })

    event_id += 1

    # --------------------------------
    # Apply segment behavior
    # --------------------------------

    device_factor = device_multiplier[row.device]

    source_factor = source_multiplier[row.traffic_source]

    customer_factor = customer_multiplier[row.customer_type]

    # --------------------------------
    # STAGE 2: PRODUCT VIEW
    # --------------------------------

    product_view_probability = (
        0.75
        * device_factor
        * source_factor
        * customer_factor
    )

    if np.random.random() > product_view_probability:
        continue

    category = np.random.choice(categories)

    product = np.random.choice(
        products[category]
    )

    time2 = base_time + pd.Timedelta(
        minutes=np.random.randint(1, 10)
    )

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": time2,
        "event_type": "product_view",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": category,
        "product": product,
        "amount": 0
    })

    event_id += 1

    # --------------------------------
    # STAGE 3: ADD TO CART
    # --------------------------------

    cart_probability = (
        0.50
        * device_factor
        * source_factor
        * customer_factor
    )

    if np.random.random() > cart_probability:
        continue

    time3 = time2 + pd.Timedelta(
        minutes=np.random.randint(1, 8)
    )

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": time3,
        "event_type": "add_to_cart",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": category,
        "product": product,
        "amount": 0
    })

    event_id += 1

    # --------------------------------
    # STAGE 4: SIGNUP
    # --------------------------------

    signup_probability = (
        0.72
        * device_factor
        * customer_factor
    )

    if np.random.random() > signup_probability:
        continue

    time4 = time3 + pd.Timedelta(
        minutes=np.random.randint(1, 8)
    )

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": time4,
        "event_type": "signup",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": category,
        "product": product,
        "amount": 0
    })

    event_id += 1

    # --------------------------------
    # STAGE 5: CHECKOUT
    # --------------------------------

    checkout_probability = (
        0.80
        * device_factor
        * source_factor
        * customer_factor
    )

    if np.random.random() > checkout_probability:
        continue

    time5 = time4 + pd.Timedelta(
        minutes=np.random.randint(1, 8)
    )

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": time5,
        "event_type": "checkout",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": category,
        "product": product,
        "amount": 0
    })

    event_id += 1

    # --------------------------------
    # STAGE 6: PURCHASE
    # --------------------------------

    purchase_probability = (
        0.65
        * device_factor
        * source_factor
        * customer_factor
    )

    if np.random.random() > purchase_probability:
        continue

    time6 = time5 + pd.Timedelta(
        minutes=np.random.randint(1, 8)
    )

    low, high = price_ranges[category]

    amount = round(
        np.random.uniform(
            low,
            high
        ),
        2
    )

    events.append({
        "event_id": f"E{event_id:08d}",
        "user_id": row.user_id,
        "session_id": row.session_id,
        "event_timestamp": time6,
        "event_type": "purchase",
        "device": row.device,
        "location": row.location,
        "traffic_source": row.traffic_source,
        "customer_type": row.customer_type,
        "product_category": category,
        "product": product,
        "amount": amount
    })

    event_id += 1


print("Events generated:", len(events))
```
