```python
valid_events = [
    "visit",
    "product_view",
    "add_to_cart",
    "signup",
    "checkout",
    "purchase"
]

invalid_events = df[
    ~df["event_type"].isin(valid_events)
]

print(
    "Invalid events:",
    len(invalid_events)
)
```

```python
print("Devices:")
print(df["device"].unique())

print("\nLocations:")
print(df["location"].unique())

print("\nTraffic Sources:")
print(df["traffic_source"].unique())

print("\nCustomer Types:")
print(df["customer_type"].unique())
```

```python
df["amount"].describe()
```

```python
df.groupby("event_type")["amount"].sum()
```

```python
invalid_purchases = df[
    (df["event_type"] == "purchase") &
    (df["amount"] <= 0)
]

print(
    "Invalid purchase records:",
    len(invalid_purchases)
)
```

```python
invalid_non_purchase = df[
    (df["event_type"] != "purchase") &
    (df["amount"] != 0)
]

print(
    "Invalid non-purchase amounts:",
    len(invalid_non_purchase)
)
```

```python
df["date"] = df["event_timestamp"].dt.date

df["year"] = df["event_timestamp"].dt.year

df["month"] = df["event_timestamp"].dt.month

df["month_name"] = df["event_timestamp"].dt.month_name()

df["week"] = df["event_timestamp"].dt.isocalendar().week

df["day_of_week"] = df["event_timestamp"].dt.day_name()

df["hour"] = df["event_timestamp"].dt.hour

df["is_weekend"] = (
    df["event_timestamp"].dt.dayofweek >= 5
).astype(int)
```

```python
df[
    [
        "event_timestamp",
        "date",
        "year",
        "month",
        "month_name",
        "week",
        "day_of_week",
        "hour",
        "is_weekend"
    ]
].head()
```

```python
df["date"]
```

```python
df["event_date"] = (
    df["event_timestamp"]
    .dt.normalize()
)
```

```python
df["event_date"].head()
```

```python
df = df.sort_values(
    by=[
        "session_id",
        "event_timestamp"
    ]
).reset_index(drop=True)
```

```python
event_order = {
    "visit": 1,
    "product_view": 2,
    "add_to_cart": 3,
    "signup": 4,
    "checkout": 5,
    "purchase": 6
}

df["event_order"] = df["event_type"].map(
    event_order
)
```

```python
df[
    [
        "session_id",
        "event_type",
        "event_order"
    ]
].head(20)
```

```python
sequence_check = (
    df.groupby("session_id")["event_order"]
    .apply(lambda x: x.is_monotonic_increasing)
)

print(
    "Sessions with invalid event sequence:",
    (~sequence_check).sum()
)
```

```python
session_duration = (
    df.groupby("session_id")["event_timestamp"]
    .agg(
        session_start="min",
        session_end="max"
    )
)

session_duration["duration_minutes"] = (
    (
        session_duration["session_end"]
        - session_duration["session_start"]
    ).dt.total_seconds()
    / 60
)

session_duration.head()
```

```python
df = df.merge(
    session_duration[
        ["duration_minutes"]
    ],
    left_on="session_id",
    right_index=True,
    how="left"
)
```

```python
df[
    [
        "session_id",
        "event_type",
        "duration_minutes"
    ]
].head(10)
```

```python
purchase_users = set(
    df.loc[
        df["event_type"] == "purchase",
        "user_id"
    ]
)

df["user_purchased"] = (
    df["user_id"]
    .isin(purchase_users)
    .astype(int)
)
```

```python
df[
    [
        "user_id",
        "event_type",
        "user_purchased"
    ]
].head(20)
```

```python
print("========== DATA QUALITY SUMMARY ==========")

print("Rows:", len(df))

print("Columns:", len(df.columns))

print("Unique users:", df["user_id"].nunique())

print("Unique sessions:", df["session_id"].nunique())

print("Unique events:", df["event_id"].nunique())

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print(
    "Duplicate event IDs:",
    df["event_id"].duplicated().sum()
)

print(
    "Invalid event types:",
    (~df["event_type"].isin(valid_events)).sum()
)

print(
    "Negative amounts:",
    (df["amount"] < 0).sum()
)

print(
    "Invalid purchases:",
    (
        (df["event_type"] == "purchase") &
        (df["amount"] <= 0)
    ).sum()
)
```

```python
processed_path = "../data/processed/user_events_clean.csv"

df.to_csv(
    processed_path,
    index=False
)

print("Clean dataset saved successfully!")
```
